# Option-Implied Regime-Switching Portfolio Optimization
### Interactive Exploration & Analysis Notebook

This notebook provides an interactive walkthrough of:
1. Inferred Hidden Markov Model (HMM) volatility regimes from option data (VIX term structure, Variance Risk Premium)
2. Cleaned covariance structures via Randomized SVD
3. $L_1$-penalized portfolio allocations and cumulative returns vs. baseline strategies

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('bmh')
%matplotlib inline

# Paths
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
data_dir = os.path.join(project_root, 'data')
report_dir = os.path.join(project_root, 'report')
```

## 1. Load Macro Indicators & Inferred Regimes

```python
regimes_path = os.path.join(data_dir, '02_regimes_data.csv')
if os.path.exists(regimes_path):
    macro_df = pd.read_csv(regimes_path, index_col='Date', parse_dates=True)
    print(f"Loaded regimes data with {len(macro_df)} trading days.")
    display(macro_df.head())
else:
    print(f"File not found: {regimes_path}")
```

## 2. Inferred Regime Characteristics & VIX Term Structure

```python
if 'macro_df' in locals():
    fig, ax = plt.subplots(figsize=(14, 6))
    ax.plot(macro_df.index, macro_df['VIX_Close'], label='VIX Index', color='darkblue', lw=1.2)
    stress_periods = macro_df[macro_df['Regime'] == 1]
    ax.fill_between(macro_df.index, 0, 100, where=(macro_df['Regime'] == 1), color='crimson', alpha=0.2, label='Stress Regime (Regime 1)')
    ax.set_title('Inferred Market Regimes vs. VIX (Out-of-Sample)', fontsize=14, fontweight='bold')
    ax.set_ylabel('VIX Level')
    ax.set_ylim(0, 90)
    ax.legend(loc='upper right')
    plt.tight_layout()
    plt.show()
```
